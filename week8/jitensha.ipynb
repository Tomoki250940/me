```python
import pandas as pd
import matplotlib.pyplot as plt

# 日本語フォント（必要な場合のみ）
plt.rcParams['font.family'] = 'IPAexGothic'

df = pd.read_csv("jitensha.csv", encoding="utf-8-sig", dtype=str)

# 列名
date_col = "発生年月日（始期）"
hour_col = "発生時（始期）"

# 日付を datetime に変換
df[date_col] = pd.to_datetime(df[date_col], format="%Y%m%d", errors="coerce")

# 時間を数字に変換
df[hour_col] = pd.to_numeric(df[hour_col], errors="coerce")

# 発生日時を作成
df["発生日時"] = df[date_col] + pd.to_timedelta(df[hour_col], unit="h")

# 欠損を除外
df = df.dropna(subset=["発生日時"])

df = df[df[date_col] >= pd.to_datetime("2024-01-01")]

# ===========================
# ① 発生年月日ごとの件数
# ===========================
counts_by_date = df.groupby(df[date_col].dt.date).size()

plt.figure(figsize=(10,5))
plt.plot(counts_by_date.index, counts_by_date.values)
plt.xticks(rotation=45)
plt.xlabel("発生年月日")
plt.ylabel("事件件数")
plt.title("発生年月日別 事件件数")
plt.tight_layout()
plt.show()

# ===========================
# ② 発生時間（0〜23時）の件数
# ===========================
counts_by_hour = df.groupby(hour_col).size()

plt.figure(figsize=(8,5))
plt.bar(counts_by_hour.index, counts_by_hour.values)
plt.xlabel("発生時間（時）")
plt.ylabel("事件件数")
plt.title("時間帯別 事件件数")
plt.tight_layout()
plt.show()
```

今回作ったのは発生日時と発生時間のグラフです。
