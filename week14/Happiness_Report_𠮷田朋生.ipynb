# 2.手始めに

幸福度がどのようなものか見ていきましょう

```python
yearly_happiness = df.groupby("year")["Happiness Score"].mean()

plt.plot(yearly_happiness.index, yearly_happiness.values)
plt.xlabel("Year")
plt.ylabel("Average Happiness Score")
plt.title("Change in Happiness Score Over Time")
plt.show()
```

2022年まで反映されないですね

絞ってやってみましょう

```python
df = df[(df["year"] >= 2015) & (df["year"] <= 2022)]

yearly_happiness = df.groupby("year")["Happiness Score"].mean()

plt.plot(yearly_happiness.index, yearly_happiness.values)
plt.xlabel("Year")
plt.ylabel("Average Happiness Score")
plt.title("Change in Happiness Score (2015–2022)")
plt.show()
```

絞ったけどうまくいきませんね

チャッピーに聞きますか

```python
yearly_happiness = df.groupby("year")["Happiness Score"].mean()

plt.plot(yearly_happiness.index, yearly_happiness.values)
plt.xlabel("Year")
plt.ylabel("Average Happiness Score")
plt.title("Change in Happiness Score (2015–2022)")
plt.xticks(yearly_happiness.index)
plt.show()
```

うまくいきましたね

これで分かったのは16年以降のデータがないことですね

次に幸福度と健康の関係性について図にしてみますか

```python
plt.scatter(df["Health (Life Expectancy)"], df["Happiness Score"])
plt.xlabel("Health (Life Expectancy)")
plt.ylabel("Happiness Score")
plt.title("Happiness Score vs Health (Life Expectancy)")
plt.show()
```

見た感じ幸福度が高いと健康寿命も高そうですね

正の相関が強いことがわかります

世界と比べて日本は長寿の国と言われてるので日本をピックアップしてみましょう

```python
japan = df[df["Country"] == "Japan"]
others = df[df["Country"] != "Japan"]

plt.scatter(
    others["Health (Life Expectancy)"],
    others["Happiness Score"],
    alpha=0.5
)

plt.scatter(
    japan["Health (Life Expectancy)"],
    japan["Happiness Score"],
    s=100,
    label="Japan"
)

plt.xlabel("Health (Life Expectancy)")
plt.ylabel("Happiness Score")
plt.title("Happiness Score vs Health (Life Expectancy)")
plt.legend()
plt.show()
```

うまく日本だけを抽出できましたね

日本は幸福度は真ん中よりちょっと上くらいだけど健康寿命はとても高いですね

GDPと幸福度についても図にしてみますか

```python
plt.scatter(df["Economy (GDP per Capita)"], df["Happiness Score"])
plt.xlabel("Economy (GDP per Capita)")
plt.ylabel("Happiness Score")
plt.title("Happiness Score vs Economy (GDP per Capita)")
plt.show()
```

うまくできました。

これも日本を強調しましょう

```python
japan = df[df["Country"] == "Japan"]
others = df[df["Country"] != "Japan"]

plt.scatter(
    others["Economy (GDP per Capita)"],
    others["Happiness Score"],
    alpha=0.5
)

plt.scatter(
    japan["Economy (GDP per Capita)"],
    japan["Happiness Score"],
    s=120,
    label="Japan"
)

plt.xlabel("Economy (GDP per Capita)")
plt.ylabel("Happiness Score")
plt.title("Happiness Score vs Economy (GDP per Capita)")
plt.legend()
plt.show()
```

これも健康寿命と同じく幸福度は決して高くはないが横軸のGDPの値は高いですね
